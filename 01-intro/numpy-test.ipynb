```python
def vector_multiplication(a, b):
    assert a.shape[0] == b.shape[0]

    m = a.shape[0]
    mul = 0.0

    for i in range(m):
        mul = mul + a[i] * b[i]

    return mul
```

```python
val1 = np.arange(4) + 1
val1
```

```python
val2 = (np.arange(4) + 1) * 2
val2
```

```python
vector_multiplication(val1, val2)
```

```python
val1.dot(val2)
```

```python
def matrix_vector_multiplication(a,b):
    assert a.shape[1] == b.shape[0]

    m = a.shape[0]
    result = np.zeros(m)

    for i in range(m):
        result[i] = vector_multiplication(a[i], b)

    return result
```

```python
U = np.random.randint(low=1, high=4, size=(3,4))
U
```

```python
v = np.arange(4) + 1
v
```

```python
matrix_vector_multiplication(U, v)
```

```python
U @ v
```

```python
def matrix_matrix_multiplication(a,b):
    assert a.shape[1] == b.shape[0]

    num_rows = a.shape[0]
    num_cols = b.shape[1]
    result = np.zeros((num_rows, num_cols))

    for i in range(num_rows):
        result[:, i] = matrix_vector_multiplication(a, b[:, i])

    return result
```

```python
W = np.random.randint(low=1, high=4, size=(3,4))
W
```
